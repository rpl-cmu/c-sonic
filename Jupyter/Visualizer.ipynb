```python
import os
import sys
from pathlib import Path

# Find the repository root: the first directory at or above the working
# directory that holds both CSONIC/ and config.py. When the kernel starts
# somewhere else, run %cd /path/to/c-sonic in a cell before this one.
parent_dir = None
for candidate in [Path.cwd(), *Path.cwd().parents]:
    if (candidate / 'CSONIC').is_dir() and (candidate / 'config.py').is_file():
        parent_dir = str(candidate)
        break
if parent_dir is None:
    raise RuntimeError("Start Jupyter inside the c-sonic checkout (or %cd into it) so the repository root can be found")
if parent_dir not in sys.path:
    sys.path.insert(0, parent_dir)

import config
from CSONIC.csonic_model import CSONICModel  # the model the viewer drives
from Jupyter.functions import Visualization

args = config.get_args([])

args.datadir = os.path.join(parent_dir, 'samples', 'csonic_sample')  # dataset root that holds logs/pairs_val*.txt; see extract_sample.py
args.ckpt_path = os.path.join(parent_dir, 'pretrained', 'CSONIC_pretrained.pth')
args.superpoint_weights = os.path.join(parent_dir, 'pretrained', 'superpoint_v1.pth')
args.pretrained = 0  # the checkpoint holds every weight
args.phase = 'test'
args.batch_size = 1

sample = Visualization(args)
```
